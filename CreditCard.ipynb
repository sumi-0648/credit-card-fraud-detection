```python
#import library
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split,cross_validate,cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
#loading dataset
df=pd.read_csv("creditcard_clean.csv")
df.head()
```

```python
#basic info
df.info()
```

```python
#find null values
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#class balance (fraud is very rare)
df["Class"].value_counts()
```

```python
x=df.drop("Class",axis=1)
y=df["Class"]
```

```python
#spliting the dataset for training and testing
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.3,
    random_state=40,
    stratify=y
)
```

```python
print("Training Dataset",x_train.shape)
print("Testing Dataset: ",x_test.shape)
```

```python
#Model Training
LR=LogisticRegression(max_iter=1000)
LR.fit(x_train,y_train)
```

```python
#Testing Accuracy
y_pred=LR.predict(x_test)
accuracy=accuracy_score(y_pred,y_test)
print("Testing Accuracy:",accuracy)
```

```python
#training accuracy
y_pred_train=LR.predict(x_train)
accuracy_train=accuracy_score(y_pred_train,y_train)
print("Training Accuracy:",accuracy_train)
```

```python
#validation Accuracy
val_accuracy=cross_val_score(LR,
    x_train,y_train,cv=5
)
print("Validation Accuracy: ",val_accuracy.mean())
```

```python
#Logistic Regression - fraud detection report
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
#Random Forest Classifier
RF=RandomForestClassifier(
    n_estimators=200,max_depth=10,min_samples_leaf=10,verbose=1,n_jobs=-1,random_state=42
)
```

```python
RF.fit(x_train,y_train)
```

```python
#Testing Accuracy for Random Forest Classifier
y_pred_rf=RF.predict(x_test)
accuracy_rf=accuracy_score(y_pred_rf,y_test)
print("Testing Accuracy:",accuracy_rf)
```

```python
# Training Accuracy for Random Forest Classifier
y_pred_train_rf=RF.predict(x_train)
accuracy_train_rf=accuracy_score(y_pred_train_rf,y_train)
print("Training Accuracy:",accuracy_train_rf)
```

```python
#Random Forest - fraud detection report
print(confusion_matrix(y_test,y_pred_rf))
print(classification_report(y_test,y_pred_rf))
```

```python
#Gradient Boosting (dataset is big, so this cell takes a few minutes)
gb=GradientBoostingClassifier(
    n_estimators=100,min_samples_leaf=10,learning_rate=0.1,random_state=42
)
gb.fit(x_train,y_train)
```

```python
#Testing Accuracy of Gradient Boosting
y_pred_gb=gb.predict(x_test)
accuracy_gb=accuracy_score(y_pred_gb,y_test)
print("Testing Accuracy:",accuracy_gb)
```

```python
# Training Accuracy for Gradient Boosting
y_pred_train_gb=gb.predict(x_train)
accuracy_train_gb=accuracy_score(y_pred_train_gb,y_train)
print("Training Accuracy:",accuracy_train_gb)
```

```python
#validation accuracy (cv=3 to save time on the big dataset)
valid_gb=cross_val_score(
    gb,x_train,y_train,cv=3,n_jobs=-1
)
print("Validation Accuracy: ",valid_gb.mean())
```
